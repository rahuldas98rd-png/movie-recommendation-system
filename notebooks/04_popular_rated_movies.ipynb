# Calculate a weighted rating
WR = (v / (v+m)) x R + (m / (v+m)) x C
* v - number of votes for a movie
* m - minimum number of votes required
* R - average rating of a movie
* C - avaerage across all movies

### For calculating 'm' we are considering 90 percentile

```python
m = df['vote_count'].quantile(0.9).round(0)
print(m)
```

### Calculating 'C'

```python
C = df['vote_average'].mean().round(1)
print(C)
```

### Storing 'v' & 'R'

```python
v = df['vote_count']
print(v)
```

```python
R = df['vote_average']
print(R)
```

### Calculate weighted rating

```python
df["weighted_rating"] = ((v / (v + m)) * R) + ((m / (v + m)) * C)
```

```python
df.head(2)
```

```python
# Sorting the data on the basis of weighted rating
df.sort_values(by='weighted_rating', ascending=False, inplace=True)
df.reset_index(drop=True, inplace=True)
```
